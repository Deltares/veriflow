## Run imports and set-up logging

```python
%load_ext autoreload
%autoreload 2

import logging
import sys
import warnings
from pathlib import Path

from dotenv import load_dotenv

from veriflow import run_pipeline
from veriflow.constants import VERSION

# Append project root (parent of notebook folder) to path
sys.path.append(str(Path("..").resolve()))

from tree_plots import (
    reanalysis_timeseries_plot,
)

load_dotenv(dotenv_path="tutorial.env", override=True)

base_config = Path("config")
base_config.exists()

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    handlers=[logging.StreamHandler()],
)
logging.info(f"Running Veriflow version {VERSION}")
```

## Evaluating the results in the _veriflow_ output `DataTree`


A good starting point for "eyeball" verification is simple: just looking at your observations and re-analysis in a visual way. Use the interactive elements in the plots below to zoom, pan and compare the results of our 2 re-analysis products.

### 1 - Check out the stations in the `DataTree`


```python
# Get the aligned input for the first verification pair
dt.veriflow.get_aligned_input(dt.veriflow.verification_pairs[0])
```

```python
# Alternatively, you can access the same data  aligned input directly from the datatree
dt["PC_RDPA/aligned_input"]
```

```python
# List the stations
ds = dt.veriflow.get_aligned_input(dt.veriflow.verification_pairs[0])["observations"].to_dataset()
stations = ds.coords["station"].values
print(f"Stations: {stations}")
```

### 2 - Visualize the precipitation (per station)


```python
reanalysis_timeseries_plot(dt, station=stations[5])
```

### 3 - Visualize the precipitation (per station) and add the absolute error to the plot as well


```python
reanalysis_timeseries_plot(dt, station=stations[4], error_var="mae", error_score_name="continuous_scores")
```
